```python
# ============================================================
# CÉLULA 1: IMPORTAÇÃO DAS BIBLIOTECAS E LEITURA DOS DADOS
# ============================================================

# Importa a biblioteca pandas (manipulação de tabelas/DataFrames) com o apelido "pd"
import pandas as  pd
# Importa a biblioteca numpy (cálculos numéricos e arrays) com o apelido "np"
import numpy as np

# Lê o arquivo CSV de vendas e guarda em um DataFrame chamado df_vendas
# - encoding="utf-8": codificação do arquivo, necessária para ler acentos (ex.: "São Paulo")
# - low_memory=False: lê o arquivo inteiro de uma vez para inferir os tipos das colunas
#   com precisão (evita avisos de tipos mistos em arquivos grandes)
# - sep=",": o separador de colunas do CSV é a vírgula
df_vendas = pd.read_csv("vendas_tech.csv", encoding="utf-8", low_memory=False, sep=",")
# display() mostra o DataFrame formatado como tabela no notebook (100100 linhas x 8 colunas)
display(df_vendas)
# CÓDIGO COMENTADO: leria a planilha Excel com os gerentes das lojas (não está sendo usado ainda).
# Atenção: o nome do arquivo está com erro de digitação ("gerantes"); o arquivo real na pasta
# se chama "gerentes_lojas.xlsx". Se for descomentar, corrija o nome.
# pd.read_excel("gerantes_lojas.xlsx")
```

```python
# ============================================================
# CÉLULA 3: INSPEÇÃO INICIAL DOS DADOS
# Objetivo: entender o formato, os tipos e a qualidade dos dados antes de tratá-los
# ============================================================

# inspeção

# CÓDIGO COMENTADO: mostraria o DataFrame inteiro (já exibido na primeira célula)
# display(df_vendas)
# CÓDIGO COMENTADO: mostraria só as 15 primeiras linhas
# display(df_vendas.head(15))
# CÓDIGO COMENTADO: mostraria só as 15 últimas linhas
# display(df_vendas.tail(15))
# CÓDIGO COMENTADO: mostraria 15 linhas aleatórias (boa para ter uma visão geral sem viés de ordem)
# display(df_vendas.sample(15))

# Mostra (nº de linhas, nº de colunas) -> (100100, 8)
display(df_vendas.shape)
# Mostra a lista com o nome de todas as colunas do DataFrame
display(list(df_vendas.columns))
# Mostra o resumo técnico: tipo de cada coluna, quantidade de valores não nulos e uso de memória
# (aqui vemos que "Loja" tem valores nulos e "Data_Base" tem apenas 1 valor preenchido)
display(df_vendas.info())
# Mostra estatísticas das colunas numéricas: contagem, média, desvio padrão, mínimo, quartis e máximo
display(df_vendas.describe())
```

```python
# ============================================================
# CÉLULA 4: TRATAMENTO E LIMPEZA DOS DADOS
# ============================================================

#tratamento de dados

#colunas
# Seleciona uma única coluna (retorna uma Series); como não é atribuída a nada,
# o resultado só seria exibido, não altera o DataFrame
df_vendas["Loja"]
# CÓDIGO COMENTADO: selecionaria duas colunas de uma vez (usando lista dentro dos colchetes)
# df_vendas[["Loja", "Cliente"]]
# Cria um novo DataFrame contendo apenas as colunas "Loja" e "Cliente"
df_novo = df_vendas[["Loja", "Cliente"]]

# Cria df_analise a partir de df_vendas, removendo a coluna "Data_Base"
# (tem só 1 valor preenchido, então não serve para a análise)
df_analise = df_vendas.drop(columns=['Data_Base'])

#apagar valores nulos

# CÓDIGO COMENTADO: removeria todas as linhas que tivessem qualquer valor nulo.
# Foi descartado porque perderíamos as vendas sem loja (~2000 linhas).
# df_analise = df_analise.dropna()
# Em vez de apagar, preenche as lojas nulas com "online" (vendas sem loja física)
df_analise["Loja"] = df_analise["Loja"].fillna("online")
# Exibe o DataFrame (nesta posição o resultado não é mostrado, pois a célula só exibe a última linha)
df_analise

#tipos de dados

# Converte a coluna "Data" de texto (string) para o tipo datetime, no formato ano-mês-dia,
# permitindo operações com datas (filtros por período, extrair mês/ano etc.)
df_analise["Data"] = pd.to_datetime(df_analise["Data"], format="%Y-%m-%d")

#Padronização
# Remove espaços em branco no início e no fim dos nomes das lojas
df_analise["Loja"] = df_analise["Loja"].str.strip()
# Padroniza a capitalização: "RIO DE JANEIRO", "rio de janeiro" e "Rio de Janeiro"
# viram todos "Rio De Janeiro" (primeira letra de cada palavra maiúscula)
df_analise["Loja"] = df_analise["Loja"].str.title()


#Remover duplicatas

# Remove linhas duplicadas considerando a combinação ID_Pedido + Loja.
# keep="first" mantém a primeira ocorrência e descarta as repetidas (100100 -> 100000 linhas)
df_analise = df_analise.drop_duplicates(subset=["ID_Pedido", "Loja"], keep="first")

# Mostra o resumo do DataFrame já tratado: sem nulos e com "Data" como datetime
df_analise.info()

# Exibe o DataFrame tratado (última linha da célula, por isso é mostrado)
df_analise
```
